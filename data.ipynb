```python
import pandas as pd
```

```python
print(" تحميل بيانات الطيران")
df_header = pd.read_csv('flight_data_2018_to_2022.csv', nrows=1)
available_cols = df_header.columns.tolist()
```

```python
desired_cols = [
    'FlightDate', 'Month', 'DayOfWeek', 'Operating_Airline', 'Airline',
    'Origin', 'Dest', 'CRSDepTime', 'DepDelay', 'DepDelayMinutes',
    'DepDel15', 'Cancelled', 'Diverted', 'Distance'
]
```

```python
valid_cols = [col for col in desired_cols if col in available_cols]

print(f"تم العثور على {len(valid_cols)} عمود مطابق من القائمة.")
```

```python
df_flights = pd.read_csv('flight_data_2018_to_2022.csv', usecols=valid_cols, low_memory=False)
```

```python
if 'Cancelled' in df_flights.columns:
    df_flights = df_flights[df_flights['Cancelled'] == 0]
    df_flights.drop(columns=['Cancelled'], inplace=True, errors='ignore')
```

```python
if 'Diverted' in df_flights.columns:
    df_flights = df_flights[df_flights['Diverted'] == 0]
    df_flights.drop(columns=['Diverted'], inplace=True, errors='ignore')
```

```python
target_col = 'DepDel15' if 'DepDel15' in df_flights.columns else 'DepDelayMinutes'
if target_col in df_flights.columns:
    df_flights.dropna(subset=[target_col], inplace=True)
```

```python
print("تم التنظيف بنجاح!")
print("عدد الصفوف النظيفة:", len(df_flights))
print("الأعمدة المحملة:", df_flights.columns.tolist())
df_flights.head()
```
