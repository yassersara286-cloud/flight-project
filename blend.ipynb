```python
import pandas as pd
```

```python
print(" تحميل بيانات الطيران")
use_cols = [
    'FlightDate', 'Month', 'DayOfWeek', 'Operating_Airline',
    'Origin', 'Dest', 'CRSDepTime', 'DepDelay', 'DepDelayMinutes',
    'DepDel15', 'Cancelled', 'Diverted', 'Distance'
]
```

```python
df_flights = pd.read_csv('flight_data_2018_to_2022.csv', usecols=lambda x: x in use_cols, low_memory=False)
```

```python
if 'Cancelled' in df_flights.columns:
    df_flights = df_flights[df_flights['Cancelled'] == 0]
if 'Diverted' in df_flights.columns:
    df_flights = df_flights[df_flights['Diverted'] == 0]
```

```python
df_flights['DEP_HOUR'] = (df_flights['CRSDepTime'] // 100).astype(int)
```

```python
print(" تحميل بيانات الطقس")
df_weather = pd.read_csv('jfk_weather_cleaned.csv')
```

```python
df_weather['DATE'] = pd.to_datetime(df_weather['DATE'])
df_weather['WEATHER_MONTH'] = df_weather['DATE'].dt.month
df_weather['WEATHER_HOUR'] = df_weather['DATE'].dt.hour
```

```python
weather_summary = df_weather.groupby(['WEATHER_MONTH', 'WEATHER_HOUR']).mean(numeric_only=True).reset_index()
```

```python
print(" دمج الداتا ")
df_merged = pd.merge(
    df_flights,
    weather_summary,
    left_on=['Month', 'DEP_HOUR'],
    right_on=['WEATHER_MONTH', 'WEATHER_HOUR'],
    how='left'
)
```

```python
df_merged.fillna(df_merged.median(numeric_only=True), inplace=True)
```

```python
df_merged.to_csv('merged_flight_weather_data.csv', index=False)
```
